```python
class Mashina:
    def __init__(self,rusumi,yili,rangi,yurgani,xolati,narxi):
        self.rusumi=rusumi
        self.yili=yili
        self.rangi=rangi
        self.yurgani=yurgani
        self.xolati=xolati
        self.narxi=narxi
    def malumot(self):
        print(f"Rusumi:{self.rusumi}")
        print(f"Yili:{self.yili}")     
        print(f"Rangi:{self.rangi}")
        print(f"Yurgani:{self.yurgani}")
        print(f"Xolati:{self.xolati}")
        print(f"Narxi:{self.narxi}")
mashina1=Mashina(
    'Cobalt',
    '2024',
    'Oq',
    '12000 km',
    'Yaxshi',
    '10000$'
)
mashina1.malumot()
```
