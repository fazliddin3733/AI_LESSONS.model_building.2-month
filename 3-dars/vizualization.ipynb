```python
import pandas as pd
```

```python
df=pd.read_csv('medical_regression.csv')
```

```python
df.isnull().sum()
```

```python
def missval(df):
    for col in df.columns:
        if df[col].isnull().any():
            if df[col].dtype=='str':
                df[col]=df[col].fillna(df[col].mode()[0])
            else:
                df[col]=df[col].fillna(df[col].mean())
    return df
```

```python
df=missval(df)
df.isnull().sum()
```

```python
import matplotlib.pyplot as plt

data = df.mean(numeric_only=True)

fig, ax = plt.subplots(layout="constrained")

ax.bar(data.index, data.values)

ax.tick_params("x", rotation=45, rotation_mode="anchor")

ax.set_ylabel("Average value")
ax.set_xlabel("Features")
ax.set_title("Medical Dataset")

plt.show()
```

```python
import plotly.express as px

data = df.groupby('age')['diabetes_progression'].mean().reset_index()

fig = px.line(
    data,
    x='age',
    y='diabetes_progression',
    title='Age va Diabetes Progression o‘zgarishi'
)

fig.show()
```
