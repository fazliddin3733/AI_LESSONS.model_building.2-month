```python
import pandas as pd
```

```python
df=pd.read_csv('regression_building_energy.csv')
df.head()
```

```python
df.isnull().sum()
```

```python
def missval(df):
    for col in df.columns:
        if df[col].isnull().any():
            if df[col].dtype=='str':
                df[col]=df[col].fillna(df[col].mode()[0])
            else:
                df[col]=df[col].fillna(df[col].mean())
    return df
```

```python
df=missval(df)
df.isnull().sum()
```

```python
df.info()
```

```python
from sklearn.preprocessing import LabelEncoder
encoder=LabelEncoder()
def encodla(df):
    for col in df.columns:
        if df[col].dtype=='str':
            dummies=pd.get_dummies(df[col],prefix=col)
            df=pd.concat([df.drop(columns=col),dummies],axis=1)
        else:
            df[col]=encoder.fit_transform(df[col])
    return df
```

```python
df=encodla(df)
df.info()
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler=MinMaxScaler()
num_col=df.select_dtypes(include=['int64','float64']).columns.drop('monthly_energy_kwh')
df[num_col]=scaler.fit_transform(df[num_col])
```

```python
df.head()
```

```python
from sklearn.model_selection import train_test_split
x=df.drop('monthly_energy_kwh',axis=1)
y=df['monthly_energy_kwh']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
lr
```

```python
lr.fit(x_train,y_train)
```

```python
y_pred=lr.predict(x_test)
y_pred
```

```python
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)
```

```python
print("MAE:", mae),
print("MSE:", mse),
print("R2 Score:", r2)
```

```python
from sklearn.model_selection import KFold,cross_val_score
kf=KFold(n_splits=5,shuffle=True,random_state=42)
scores=cross_val_score(lr,x,y,cv=kf,scoring='r2')
print(scores)
```

```python
import numpy as np
np.mean(scores)
```
