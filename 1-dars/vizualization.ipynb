```python
import pandas as pd
```

```python
df=pd.read_csv('houseprice.csv')
```

```python
def missval(df):
    for col in df.columns:
        if df[col].isnull().any():
            if df[col].dtype=='str':
                df[col]=df[col].fillna(df[col].mode()[0])
            else:
                df[col]=df[col].fillna(df[col].mean())
    return df
```

```python
df=missval(df)
df.info()
```
