```python
import pandas as pd
```

```python
df=pd.read_csv('Marketing customer dataset.csv')
```

```python
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder,MinMaxScaler
class DataPreprocessing:
    def __init__(self,df):
        self.df=df
    def fillmiss(self):
        for col in self.df.columns:
            if self.df[col].isnull().any():
                if self.df[col].dtype=='str':
                    self.df[col] =self.df[col].fillna(self.df[col].mode()[0])                
                else:
                    self.df[col] = self.df[col].fillna(self.df[col].mean())
        return self
    def encodla(self):
        encoder=LabelEncoder()
        for col in self.df.columns:
            if self.df[col].dtype=='str':
                if self.df[col].nunique()<=5:
                    dummies=pd.get_dummies(self.df[col],prefix=col,dtype=int)
                    self.df=pd.concat([self.df.drop(columns=col),dummies],axis=1)
                else:
                    self.df[col]=encoder.fit_transform(self.df[col])
        return self
    def scale(self):
        scaler=MinMaxScaler()
        num_col=self.df.select_dtypes(include=['int64','float64']).columns.drop('churn')
        self.df[num_col]=scaler.fit_transform(self.df[num_col])
        return self
                    
```

```python
dp=DataPreprocessing(df)
dp.fillmiss().encodla().scale()
preprocessed_data=dp.df
```
